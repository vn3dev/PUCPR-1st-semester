- Letras maiusculas para definir o nome
- Não necessariamente precisam ser numeros
- Pra definir conjuntos usamos {}

```python
A = {1, 2, 3, 4, 8, 5}

print(A, type(A))
```

### Conjunto vazio: set() 

Podemos fazer operações com conjuntos, lembrar de não repetir elementos em caso de união U

{1, 2} U {2, 3} = {1, 2, 3}

Em python, união é representada por |

```python
A = {2, 3, 4, 6, 7, 8}
B = {2, 3, 5, 9}

print(A | B)
```

```python
A = set()

print(A, type(A))
```

Interseção: Apenas os elementos que pertencem aos dois conjuntos simultaneamente

{1, 2, 3} Interseção {2, 3, 4} = {2, 3}

Em python a interseção é representada por &

```python
A = {1, 2, 3}
B = {2, 3, 4}

print(A & B)
```

Diferença: Precisa pertencer a A, e não pertencer a B

{1, 2, 3} - {2} = {1, 3}

Em python, diferença é representada por -

```python
A = {1, 2, 3}
B = {2}

print(A - B)
```

Diferença Simétrica: Todos os elementos que estão em A ou estão em B, mas não estou em ambos (ELEMENTOS EXCLUSIVOS)

Em python, Diferença simétrica é representada por ^

Também pode ser matematicamente representado por (A união B) diferença (A interseção B)

Subconjunto: Quando um conjunto inteiro esta contido em outro conjunto

Em python, subconjunto é representado por <=

Complemento: O que falta em B para se tornar A? Equivalente a A - B

Em python, não tem complemento, usar diferença quando precisar

```python
pares = {x for x in range(1,22) if x % 2 == 0}
print(pares)
```

```python
pares = {x for x in range(1,101) if x % 2 == 0}
mult3 = {x for x in range(1,101) if x % 3 == 0}

print(pares & mult3)
```

```python
A = {8, 9, 10, 11}
B = {2, 3, 5, 8}
C = {2, 4, 6, 8, 10}

resultado = (A | B) - C
print(resultado)
```

Clientes que compraram pelo menos um produto

Clientes que compraram ambos os produtos

Clientes que não estão na base de clientes mas aparecem em compras

```python
C = {101, 102, 103, 104, 105, 106, 107, 108}

B = {102, 104, 106, 108}

A = {101, 102, 103, 105, 108}

print("Clientes que compraram pelo menos um produto:", (C & A) | (C & B))
print("Clientes que compraram ambos os produtos", (B & A))
print("Clientes que compraram apenas o produto A", A)
print("Clientes que não compraram nenhum produto", C - (A | B))
print("Clientes que compraram A mas nao B", B - A)
print("Clientes que não estão na base de clientes mas aparecem em compras", (A | B) - C)
```
